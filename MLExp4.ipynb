## Relevant Theory

### Logistic Regression

Logistic Regression is a supervised machine learning algorithm used for **binary classification**. In this experiment, it is used to classify credit card transactions into two classes:

- **0 $\rightarrow$ Legitimate transaction**
- **1 $\rightarrow$ Fraudulent transaction**

The model first calculates a linear combination of input features:

$$
z = w^T X + b
$$

where:

- $w$ = weight vector
- $X$ = input feature vector
- $b$ = bias/intercept
- $z$ = linear predictor

---

### Sigmoid Function

The Sigmoid Function converts the value of $z$ into a probability between 0 and 1:

$$
\sigma(z) = \frac{1}{1 + e^{-z}}
$$

The predicted probability that a transaction is fraudulent is:

$$
P(y=1|X) = \sigma(w^T X + b)
$$

The predicted class can be obtained using a threshold:

$$
\hat{y} =
\begin{cases}
1, & \text{if } P(y=1|X) \geq 0.5 \\
0, & \text{if } P(y=1|X) < 0.5
\end{cases}
$$

---
## Cost Function

Logistic Regression uses the **Binary Cross-Entropy Loss** or **Log Loss**.

The cost function is:

$$
J(w,b) =
-\frac{1}{m}
\sum_{i=1}^{m}
\left[
y_i \log(\hat{y}_i)
+
(1-y_i)\log(1-\hat{y}_i)
\right]
$$

where:

- $m$ = number of training examples
- $y_i$ = actual class label
- $\hat{y}_i$ = predicted probability

The objective of training is to find values of $w$ and $b$ that minimize the cost function.

---

## Gradient Descent

Gradient Descent is used to optimize the parameters of the Logistic Regression model.

The weights are updated using:

$$
w := w - \alpha \frac{\partial J}{\partial w}
$$

The bias is updated using:

$$
b := b - \alpha \frac{\partial J}{\partial b}
$$

where $\alpha$ is the learning rate.

For Logistic Regression, the gradient of the cost function with respect to the weights can be expressed as:

$$
\frac{\partial J}{\partial w}
=
\frac{1}{m}X^T(\hat{y}-y)
$$

The gradient with respect to the bias is:

$$
\frac{\partial J}{\partial b}
=
\frac{1}{m}
\sum_{i=1}^{m}
(\hat{y}_i-y_i)
$$

Therefore, the parameter updates become:

$$
w := w - \alpha
\left[
\frac{1}{m}X^T(\hat{y}-y)
\right]
$$

$$
b := b - \alpha
\left[
\frac{1}{m}
\sum_{i=1}^{m}
(\hat{y}_i-y_i)
\right]
$$

---

## Class Imbalance

Credit card fraud datasets are generally highly imbalanced because legitimate transactions significantly outnumber fraudulent transactions.

For example, the dataset may contain a very large number of legitimate transactions and only a small number of fraudulent transactions.

A model that predicts most transactions as legitimate may achieve high accuracy but perform poorly at detecting fraud.

Therefore, techniques such as:

- Random Undersampling
- Oversampling
- Class Weighting

can be used to handle class imbalance.

## Evaluation Metrics

### Confusion Matrix

The confusion matrix consists of four components:

- **True Positive (TP):** Fraudulent transaction correctly identified as fraud.
- **True Negative (TN):** Legitimate transaction correctly identified as legitimate.
- **False Positive (FP):** Legitimate transaction incorrectly identified as fraud.
- **False Negative (FN):** Fraudulent transaction incorrectly identified as legitimate.

---

### Accuracy

Accuracy measures the proportion of correctly classified transactions:

$$
Accuracy =
\frac{TP+TN}
{TP+TN+FP+FN}
$$

However, accuracy alone is not sufficient for highly imbalanced fraud detection datasets.

---

### Precision

Precision indicates the proportion of transactions predicted as fraudulent that are actually fraudulent:

$$
Precision =
\frac{TP}
{TP+FP}
$$

A high Precision means that the model produces fewer false fraud alerts.

---

### Recall

Recall measures the proportion of actual fraudulent transactions correctly detected:

$$
Recall =
\frac{TP}
{TP+FN}
$$

Recall is particularly important in fraud detection because a **False Negative** represents a fraudulent transaction that was not detected.

---

### F1-Score

F1-Score is the harmonic mean of Precision and Recall:

$$
F1 =
2 \times
\frac{Precision \times Recall}
{Precision + Recall}
$$

---

## ROC Curve

The Receiver Operating Characteristic (ROC) curve represents the relationship between the **True Positive Rate (TPR)** and **False Positive Rate (FPR)** at different classification thresholds.

The True Positive Rate is:

$$
TPR =
\frac{TP}
{TP+FN}
$$

The False Positive Rate is:

$$
FPR =
\frac{FP}
{FP+TN}
$$

The ROC curve helps determine how well the model separates fraudulent transactions from legitimate transactions.

---

## *Area Under the Curve (AUC)*

The **Area Under the ROC Curve (AUC)** summarizes the performance of the classifier across different classification thresholds.

An AUC value close to:

$$
AUC = 1
$$

indicates excellent discrimination, while:

$$
AUC = 0.5
$$

indicates performance approximately equivalent to random classification.

## Task 3: Data Preprocessing

Check the dataset for the following:

* Missing values.
* Duplicate records.
* Categorical variables.
* Numerical variables.

Identify the number of missing values in each column and handle them using suitable techniques such as **removal or imputation**.

Identify categorical variables and convert them into numerical form using appropriate **encoding techniques**.

For example, categorical variables can be converted into numerical values using **One-Hot Encoding**.

Also, identify the numerical features that will be used for building the Logistic Regression model.

The objective of this task is to prepare a clean and suitable dataset for further processing, feature scaling, model training, and evaluation.


```python

# ============================================================
# TASK 3: DATA PREPROCESSING
# ============================================================


# ------------------------------------------------------------
# 1. Check Missing Values
# ------------------------------------------------------------

print("============================================")
print("1. MISSING VALUES")
print("============================================")

missing_values = df.isnull().sum()

print("\nMissing values in each column:")
print(missing_values)

print("\nTotal missing values:",
      missing_values.sum())


# ------------------------------------------------------------
# 2. Handle Missing Values
# ------------------------------------------------------------

print("\n============================================")
print("2. HANDLE MISSING VALUES")
print("============================================")

if missing_values.sum() == 0:

    print("No missing values found.")
    print("No imputation or removal is required.")

else:

    # Identify numerical columns
    numerical_columns_temp = df.select_dtypes(
        include=np.number
    ).columns

    # Fill numerical missing values using median
    for column in numerical_columns_temp:

        if df[column].isnull().sum() > 0:

            df[column] = df[column].fillna(
                df[column].median()
            )

    # Identify categorical columns
    categorical_columns_temp = df.select_dtypes(
        include=["object", "category", "bool"]
    ).columns

    # Fill categorical missing values using mode
    for column in categorical_columns_temp:

        if df[column].isnull().sum() > 0:

            df[column] = df[column].fillna(
                df[column].mode()[0]
            )

    print("Missing values have been handled.")


# Verify missing values again

print("\nMissing values after handling:")
print(df.isnull().sum())


# ------------------------------------------------------------
# 3. Check Duplicate Records
# ------------------------------------------------------------

print("\n============================================")
print("3. DUPLICATE RECORDS")
print("============================================")

duplicate_count = df.duplicated().sum()

print("Number of duplicate records:",
      duplicate_count)


# ------------------------------------------------------------
# 4. Remove Duplicate Records
# ------------------------------------------------------------

if duplicate_count > 0:

    df = df.drop_duplicates().reset_index(drop=True)

    print("Duplicate records removed.")

else:

    print("No duplicate records found.")
    print("No removal required.")


print("\nDataset shape after duplicate handling:")
print(df.shape)


# ------------------------------------------------------------
# 5. Identify Categorical Variables
# ------------------------------------------------------------

print("\n============================================")
print("4. CATEGORICAL VARIABLES")
print("============================================")

categorical_columns = df.select_dtypes(
    include=["object", "category", "bool"]
).columns.tolist()

if len(categorical_columns) == 0:

    print("No categorical variables found.")

else:

    print("Categorical variables:")
    print(categorical_columns)


# ------------------------------------------------------------
# 6. Encode Categorical Variables
# ------------------------------------------------------------

print("\n============================================")
print("5. CATEGORICAL ENCODING")
print("============================================")

if len(categorical_columns) == 0:

    print("No categorical variables to encode.")

else:

    # One-Hot Encoding
    df = pd.get_dummies(
        df,
        columns=categorical_columns,
        drop_first=True
    )

    print("Categorical variables encoded using")
    print("One-Hot Encoding.")


# ------------------------------------------------------------
# 7. Identify Numerical Variables
# ------------------------------------------------------------

print("\n============================================")
print("6. NUMERICAL VARIABLES")
print("============================================")

numerical_columns = df.select_dtypes(
    include=np.number
).columns.tolist()

print("Numerical variables:")

for column in numerical_columns:
    print(column)


# ------------------------------------------------------------
# 8. Separate Features and Target
# ------------------------------------------------------------

print("\n============================================")
print("7. FEATURES AND TARGET")
print("============================================")

target = "fraud"

X = df.drop(columns=[target])

y = df[target]

print("Feature variables:")
print(X.columns.tolist())

print("\nTarget variable:")
print(target)


# ------------------------------------------------------------
# 9. Final Dataset Check
# ------------------------------------------------------------

print("\n============================================")
print("8. FINAL PREPROCESSING CHECK")
print("============================================")

print("\nDataset shape:")
print(df.shape)

print("\nTotal missing values:")
print(df.isnull().sum().sum())

print("\nTotal duplicate records:")
print(df.duplicated().sum())

print("\nNumber of features:")
print(X.shape[1])

print("\nFeature variables:")
print(X.columns.tolist())

print("\nTarget variable:")
print(target)


# ------------------------------------------------------------
# 10. Display Final Data
# ------------------------------------------------------------

print("\n============================================")
print("9. FINAL DATASET")
print("============================================")

display(df.head())
```

## Task 4: Feature Scaling and Class Imbalance Handling

Standardize the numerical features using **Standardization** so that they have approximately zero mean and unit variance.

The standardized value of a feature is calculated as:

$$
z = \frac{x-\mu}{\sigma}
$$

where:

* (x) = original feature value
* (\mu) = mean of the feature
* (\sigma) = standard deviation of the feature
* (z) = standardized feature value

Handle the **class imbalance** in the dataset using a suitable technique such as:

* Random Undersampling
* Oversampling
* Class Weighting

Display the class distribution **before and after** applying the selected class-imbalance handling technique.

Compare the number of legitimate and fraudulent transactions before and after balancing.

The objective of this task is to ensure that the features are on a comparable scale and that the Logistic Regression model does not become biased toward the majority class.


```python
# ============================================================
# TASK 4: FEATURE SCALING AND CLASS IMBALANCE HANDLING
# ============================================================


# ------------------------------------------------------------
# 1. Separate Features and Target
# ------------------------------------------------------------

target = "fraud"

X = df.drop(columns=[target])
y = df[target]

print("Features:")
print(X.columns.tolist())

print("\nTarget:")
print(target)


# ------------------------------------------------------------
# 2. Split Dataset into Training and Testing Sets
# ------------------------------------------------------------

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("\n============================================")
print("TRAIN-TEST SPLIT")
print("============================================")

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))


# ------------------------------------------------------------
# 3. Display Class Distribution Before Balancing
# ------------------------------------------------------------

print("\n============================================")
print("CLASS DISTRIBUTION BEFORE BALANCING")
print("============================================")

print("\nTraining set:")
print(y_train.value_counts())

legitimate_before = (y_train == 0).sum()
fraud_before = (y_train == 1).sum()

print("\nLegitimate transactions:", legitimate_before)
print("Fraudulent transactions:", fraud_before)


# ------------------------------------------------------------
# 4. Standardize Numerical Features
# ------------------------------------------------------------

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)

X_test_scaled = scaler.transform(X_test)

print("\n============================================")
print("FEATURE SCALING")
print("============================================")

print("Standardization completed.")

print("\nTraining data mean:")
print(X_train_scaled.mean(axis=0))

print("\nTraining data standard deviation:")
print(X_train_scaled.std(axis=0))


# ------------------------------------------------------------
# 5. Convert Scaled Data Back to DataFrame
# ------------------------------------------------------------

X_train_scaled = pd.DataFrame(
    X_train_scaled,
    columns=X_train.columns,
    index=X_train.index
)

X_test_scaled = pd.DataFrame(
    X_test_scaled,
    columns=X_test.columns,
    index=X_test.index
)


# ------------------------------------------------------------
# 6. Random Undersampling
# ------------------------------------------------------------

print("\n============================================")
print("RANDOM UNDERSAMPLING")
print("============================================")

# Separate legitimate and fraudulent transactions

legitimate_indices = y_train[y_train == 0].index
fraudulent_indices = y_train[y_train == 1].index

# Number of fraudulent transactions
n_fraud = len(fraudulent_indices)

# Randomly select the same number of legitimate transactions
np.random.seed(42)

selected_legitimate_indices = np.random.choice(
    legitimate_indices,
    size=n_fraud,
    replace=False
)

# Combine both classes
balanced_indices = np.concatenate([
    selected_legitimate_indices,
    fraudulent_indices
])

# Shuffle the balanced indices
np.random.shuffle(balanced_indices)


# Create balanced training data

X_train_balanced = X_train_scaled.loc[balanced_indices]

y_train_balanced = y_train.loc[balanced_indices]


# ------------------------------------------------------------
# 7. Display Class Distribution After Balancing
# ------------------------------------------------------------

print("\n============================================")
print("CLASS DISTRIBUTION AFTER BALANCING")
print("============================================")

print("\nBalanced training set:")
print(y_train_balanced.value_counts())

legitimate_after = (y_train_balanced == 0).sum()
fraud_after = (y_train_balanced == 1).sum()

print("\nLegitimate transactions:", legitimate_after)
print("Fraudulent transactions:", fraud_after)


# ------------------------------------------------------------
# 8. Compare Before and After Balancing
# ------------------------------------------------------------

print("\n============================================")
print("BEFORE VS AFTER BALANCING")
print("============================================")

comparison = pd.DataFrame({
    "Class": [
        "Legitimate",
        "Fraudulent"
    ],

    "Before Balancing": [
        legitimate_before,
        fraud_before
    ],

    "After Balancing": [
        legitimate_after,
        fraud_after
    ]
})

display(comparison)


# ------------------------------------------------------------
# 9. Calculate Percentages
# ------------------------------------------------------------

before_total = legitimate_before + fraud_before
after_total = legitimate_after + fraud_after

before_legitimate_percentage = (
    legitimate_before / before_total
) * 100

before_fraud_percentage = (
    fraud_before / before_total
) * 100

after_legitimate_percentage = (
    legitimate_after / after_total
) * 100

after_fraud_percentage = (
    fraud_after / after_total
) * 100


print("\n============================================")
print("CLASS PERCENTAGES")
print("============================================")

print("\nBefore balancing:")

print(
    f"Legitimate: {before_legitimate_percentage:.2f}%"
)

print(
    f"Fraudulent: {before_fraud_percentage:.2f}%"
)


print("\nAfter balancing:")

print(
    f"Legitimate: {after_legitimate_percentage:.2f}%"
)

print(
    f"Fraudulent: {after_fraud_percentage:.2f}%"
)


# ------------------------------------------------------------
# 10. Plot Before and After Class Distribution
# ------------------------------------------------------------

fig, axes = plt.subplots(1, 2, figsize=(12, 5))


# Before balancing

axes[0].bar(
    ["Legitimate", "Fraudulent"],
    [legitimate_before, fraud_before]
)

axes[0].set_title("Before Balancing")
axes[0].set_xlabel("Transaction Class")
axes[0].set_ylabel("Number of Transactions")


# After balancing

axes[1].bar(
    ["Legitimate", "Fraudulent"],
    [legitimate_after, fraud_after]
)

axes[1].set_title("After Random Undersampling")
axes[1].set_xlabel("Transaction Class")
axes[1].set_ylabel("Number of Transactions")


plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 11. Final Dataset Information
# ------------------------------------------------------------

print("\n============================================")
print("TASK 4 COMPLETE")
print("============================================")

print(
    "Original training samples:",
    len(X_train)
)

print(
    "Balanced training samples:",
    len(X_train_balanced)
)

print(
    "Testing samples:",
    len(X_test)
)

print(
    "\nFinal balanced training class distribution:"
)

print(y_train_balanced.value_counts())
```

## Task 6: Implement Sigmoid and Cost Functions

Implement the **Sigmoid Function** used in Logistic Regression.

The Sigmoid Function is defined as:

$$
\sigma(z) = \frac{1}{1+e^{-z}}
$$

The Sigmoid Function converts the output of the linear equation into a probability value between 0 and 1.

Implement the **Binary Cross-Entropy Cost Function**:

$$
J(w,b) =
-\frac{1}{m}
\sum_{i=1}^{m}
\left[
y_i\log(\hat{y}_i)
+
(1-y_i)\log(1-\hat{y}_i)
\right]
$$

where:

* (m) = number of training samples
* (y_i) = actual class label
* (\hat{y}_i) = predicted probability
* (w) = weight vector
* (b) = bias

Test the Sigmoid Function using suitable sample values.

Calculate the initial cost of the Logistic Regression model using the initialized weights and bias.

The objective of this task is to understand how the Sigmoid Function produces probabilities and how the Binary Cross-Entropy Cost Function measures the error of the Logistic Regression model.


```python
# ============================================================
# TASK 6: IMPLEMENT SIGMOID AND COST FUNCTIONS
# ============================================================


# ------------------------------------------------------------
# 1. Sigmoid Function
# ------------------------------------------------------------

def sigmoid(z):
    """
    Sigmoid function:

        sigmoid(z) = 1 / (1 + e^(-z))
    """

    return 1 / (1 + np.exp(-z))


# ------------------------------------------------------------
# 2. Test Sigmoid Function
# ------------------------------------------------------------

print("============================================")
print("SIGMOID FUNCTION TEST")
print("============================================")

sample_values = np.array([
    -10,
    -5,
    0,
    5,
    10
])

sigmoid_values = sigmoid(sample_values)

for z, result in zip(sample_values, sigmoid_values):
    print(f"sigmoid({z}) = {result:.6f}")


# ------------------------------------------------------------
# 3. Binary Cross-Entropy Cost Function
# ------------------------------------------------------------

def compute_cost(X, y, w, b):
    """
    Binary Cross-Entropy Cost Function:

    J(w,b) = -1/m * sum[
        y*log(y_hat) +
        (1-y)*log(1-y_hat)
    ]
    """

    m = len(y)

    # Calculate linear combination
    z = np.dot(X, w) + b

    # Calculate predicted probabilities
    y_hat = sigmoid(z)

    # Prevent log(0)
    epsilon = 1e-15

    y_hat = np.clip(
        y_hat,
        epsilon,
        1 - epsilon
    )

    # Calculate Binary Cross-Entropy cost
    cost = -(1 / m) * np.sum(
        y * np.log(y_hat)
        + (1 - y) * np.log(1 - y_hat)
    )

    return cost


# ------------------------------------------------------------
# 4. Initialize Weights and Bias
# ------------------------------------------------------------

# Convert training data to NumPy arrays

X_train_array = X_train.to_numpy(dtype=float)

y_train_array = y_train.to_numpy(dtype=float)


# Initialize weights with zeros

w = np.zeros(X_train_array.shape[1])

# Initialize bias with zero

b = 0.0


print("\n============================================")
print("INITIAL PARAMETERS")
print("============================================")

print("Number of weights:", len(w))

print("Initial weights:")
print(w)

print("\nInitial bias:")
print(b)


# ------------------------------------------------------------
# 5. Calculate Initial Predictions
# ------------------------------------------------------------

z_initial = np.dot(
    X_train_array,
    w
) + b

y_hat_initial = sigmoid(z_initial)


print("\n============================================")
print("INITIAL PREDICTIONS")
print("============================================")

print("First 10 predicted probabilities:")

print(y_hat_initial[:10])


# ------------------------------------------------------------
# 6. Calculate Initial Cost
# ------------------------------------------------------------

initial_cost = compute_cost(
    X_train_array,
    y_train_array,
    w,
    b
)

print("\n============================================")
print("INITIAL COST")
print("============================================")

print(
    f"Initial Binary Cross-Entropy Cost: "
    f"{initial_cost:.6f}"
)
```

## Task 7: Implement Logistic Regression Using Gradient Descent

Implement **Logistic Regression from scratch** using NumPy without directly using the Scikit-learn Logistic Regression implementation.

Initialize the model parameters, including the **weight vector** (w) and **bias** (b).

Calculate the predicted probability using the Sigmoid Function:

$$
\hat{y} = \sigma(Xw+b)
$$

where:

* (X) = input feature matrix
* (w) = weight vector
* $\hat{y} = \text{predicted probability}$


Calculate the gradients of the cost function with respect to the weights and bias:

<div align="left">

$$
\frac{\partial J}{\partial w}
=
\frac{1}{m}X^T(\hat{y}-y)
$$

$$
\frac{\partial J}{\partial b}
=
\frac{1}{m}\sum_{i=1}^{m}(\hat{y}_i-y_i)
$$

</div>

Update the model parameters using **Gradient Descent**:

$$
w := w-\alpha\frac{\partial J}{\partial w}
$$

$$
b := b-\alpha\frac{\partial J}{\partial b}
$$

where ($\alpha$) represents the **learning rate**.

Train the model for a suitable number of iterations and record the cost value during each iteration.

Plot **Cost vs. Iteration** to verify whether the optimization process is converging.

The objective of this task is to implement Logistic Regression manually and understand how Gradient Descent optimizes the model parameters by minimizing the Binary Cross-Entropy Cost Function.


```python
# ============================================================
# TASK 7: LOGISTIC REGRESSION USING GRADIENT DESCENT
# ============================================================


# ------------------------------------------------------------
# 1. Prepare Training Data
# ------------------------------------------------------------

X_train_array = X_train.to_numpy(dtype=float)
y_train_array = y_train.to_numpy(dtype=float)

print("Training data shape:")
print(X_train_array.shape)

print("\nTraining target shape:")
print(y_train_array.shape)


# ------------------------------------------------------------
# 2. Initialize Model Parameters
# ------------------------------------------------------------

n_features = X_train_array.shape[1]

# Initialize weights with zeros
w = np.zeros(n_features)

# Initialize bias with zero
b = 0.0

print("\n============================================")
print("INITIALIZE PARAMETERS")
print("============================================")

print("Number of features:", n_features)

print("\nInitial weights:")
print(w)

print("\nInitial bias:")
print(b)


# ------------------------------------------------------------
# 3. Set Gradient Descent Parameters
# ------------------------------------------------------------

learning_rate = 0.01

iterations = 1000

print("\n============================================")
print("GRADIENT DESCENT PARAMETERS")
print("============================================")

print("Learning rate:", learning_rate)
print("Number of iterations:", iterations)


# ------------------------------------------------------------
# 4. Gradient Descent Training
# ------------------------------------------------------------

m = len(y_train_array)

cost_history = []


for i in range(iterations):

    # ----------------------------------------
    # Calculate linear prediction
    # ----------------------------------------

    z = np.dot(X_train_array, w) + b


    # ----------------------------------------
    # Calculate predicted probability
    # ----------------------------------------

    y_hat = sigmoid(z)


    # ----------------------------------------
    # Prevent numerical problems
    # ----------------------------------------

    epsilon = 1e-15

    y_hat_clipped = np.clip(
        y_hat,
        epsilon,
        1 - epsilon
    )


    # ----------------------------------------
    # Calculate Binary Cross-Entropy Cost
    # ----------------------------------------

    cost = -(1 / m) * np.sum(
        y_train_array * np.log(y_hat_clipped)
        +
        (1 - y_train_array)
        * np.log(1 - y_hat_clipped)
    )


    # ----------------------------------------
    # Calculate gradients
    # ----------------------------------------

    dw = (
        1 / m
    ) * np.dot(
        X_train_array.T,
        (y_hat - y_train_array)
    )

    db = (
        1 / m
    ) * np.sum(
        y_hat - y_train_array
    )


    # ----------------------------------------
    # Update weights and bias
    # ----------------------------------------

    w = w - learning_rate * dw

    b = b - learning_rate * db


    # ----------------------------------------
    # Store cost
    # ----------------------------------------

    cost_history.append(cost)


    # ----------------------------------------
    # Display cost periodically
    # ----------------------------------------

    if i % 100 == 0:

        print(
            f"Iteration {i}: Cost = {cost:.6f}"
        )


# ------------------------------------------------------------
# 5. Final Parameters
# ------------------------------------------------------------

print("\n============================================")
print("TRAINING COMPLETED")
print("============================================")

print("\nFinal weights:")
print(w)

print("\nFinal bias:")
print(b)

print("\nInitial cost:")
print(cost_history[0])

print("\nFinal cost:")
print(cost_history[-1])


# ------------------------------------------------------------
# 6. Plot Cost vs Iteration
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

plt.plot(
    range(1, iterations + 1),
    cost_history
)

plt.xlabel("Iteration")

plt.ylabel("Cost")

plt.title(
    "Cost vs. Iteration during Gradient Descent"
)

plt.grid(True)

plt.show()


# ------------------------------------------------------------
# 7. Check Convergence
# ------------------------------------------------------------

print("\n============================================")
print("CONVERGENCE CHECK")
print("============================================")

if cost_history[-1] < cost_history[0]:

    print(
        "Cost decreased during training."
    )

    print(
        "Gradient Descent is converging."
    )

else:

    print(
        "Cost did not decrease."
    )

    print(
        "Learning rate or model settings "
        "may need adjustment."
    )
```

## Task 8: Generate Predictions and Evaluate the Confusion Matrix

Use the trained Logistic Regression model to calculate the **fraud probabilities** for the validation and test datasets.

The predicted probability is calculated using:

$$
\hat{y} = \sigma(Xw+b)
$$

Convert the predicted probabilities into class labels using a threshold of **0.5**:

$$
\hat{y}_{class} =
\begin{cases}
1, & \text{if } \hat{y} \geq 0.5 \
0, & \text{if } \hat{y} < 0.5
\end{cases}
$$

Construct the **Confusion Matrix** and identify the following:

* **True Positive (TP):** Fraudulent transactions correctly classified as fraudulent.
* **True Negative (TN):** Legitimate transactions correctly classified as legitimate.
* **False Positive (FP):** Legitimate transactions incorrectly classified as fraudulent.
* **False Negative (FN):** Fraudulent transactions incorrectly classified as legitimate.

The confusion matrix can be represented as:

$$
\text{Confusion Matrix} =
\begin{bmatrix}
TN & FP \\
FN & TP
\end{bmatrix}
$$

Explain the meaning of each result in the context of **credit card fraud detection**.

The objective of this task is to evaluate the classification results of the Logistic Regression model and understand the different types of prediction errors, with particular attention to **False Negatives**, which represent fraudulent transactions that are not detected.


```python
# Task 8: Generate Predictions and Evaluate Confusion Matrix

import numpy as np

# Sigmoid function
def sigmoid(z):
    return 1 / (1 + np.exp(-z))


# ---------------------------------------------------------
# 1. Calculate fraud probabilities
# ---------------------------------------------------------

# Validation probabilities
y_val_prob = sigmoid(X_val @ w + b)

# Test probabilities
y_test_prob = sigmoid(X_test @ w + b)


# ---------------------------------------------------------
# 2. Convert probabilities into class labels
#    Threshold = 0.5
# ---------------------------------------------------------

y_val_pred = (y_val_prob >= 0.5).astype(int)
y_test_pred = (y_test_prob >= 0.5).astype(int)


# ---------------------------------------------------------
# 3. Function to calculate confusion matrix
# ---------------------------------------------------------

def confusion_matrix_manual(y_actual, y_pred):

    TP = np.sum((y_actual == 1) & (y_pred == 1))
    TN = np.sum((y_actual == 0) & (y_pred == 0))
    FP = np.sum((y_actual == 0) & (y_pred == 1))
    FN = np.sum((y_actual == 1) & (y_pred == 0))

    return TN, FN, FP, TP


# ---------------------------------------------------------
# 4. Validation Confusion Matrix
# ---------------------------------------------------------

TN_val, FN_val, FP_val, TP_val = confusion_matrix_manual(
    y_val, y_val_pred
)

print("=" * 55)
print("VALIDATION SET")
print("=" * 55)

print("True Negative (TN):", TN_val)
print("False Negative (FN):", FN_val)
print("False Positive (FP):", FP_val)
print("True Positive (TP):", TP_val)

print("\nConfusion Matrix:")
print(np.array([
    [TN_val, FN_val],
    [FP_val, TP_val]
]))


# ---------------------------------------------------------
# 5. Test Confusion Matrix
# ---------------------------------------------------------

TN_test, FN_test, FP_test, TP_test = confusion_matrix_manual(
    y_test, y_test_pred
)

print("\n" + "=" * 55)
print("TEST SET")
print("=" * 55)

print("True Negative (TN):", TN_test)
print("False Negative (FN):", FN_test)
print("False Positive (FP):", FP_test)
print("True Positive (TP):", TP_test)

print("\nConfusion Matrix:")
print(np.array([
    [TN_test, FN_test],
    [FP_test, TP_test]
]))
```

## Task 10: Plot ROC Curve, Calculate AUC and Interpret Results

Calculate the **False Positive Rate (FPR)** and **True Positive Rate (TPR)** for different classification thresholds.

The False Positive Rate is calculated as:

$$
FPR =
\frac{FP}{FP+TN}
$$

The True Positive Rate is equivalent to Recall and is calculated as:

$$
TPR =
\frac{TP}{TP+FN}
$$

Plot the **Receiver Operating Characteristic (ROC) Curve)** by plotting FPR on the (x)-axis and TPR on the (y)-axis for different classification thresholds.

Calculate the **Area Under the ROC Curve (AUC)** to measure the ability of the Logistic Regression model to distinguish between fraudulent and legitimate transactions.

Prepare a final performance summary containing:

* Accuracy
* Precision
* Recall
* F1-Score
* ROC-AUC
* Confusion Matrix

Interpret the model performance and discuss the trade-off between **False Positives and False Negatives**.

A higher ROC-AUC value indicates better ability of the model to distinguish between fraudulent and legitimate transactions.

Explain whether the trained Logistic Regression model is suitable as a **baseline model for credit card fraud detection**.

The objective of this task is to evaluate the overall classification performance of the Logistic Regression model using the ROC Curve and AUC and to understand the trade-off between detecting fraudulent transactions and incorrectly flagging legitimate transactions.


```python
# Task 9: Calculate Precision, Recall, F1-Score and Accuracy

# ---------------------------------------------------------
# Function to calculate classification metrics
# ---------------------------------------------------------

def calculate_metrics(TP, TN, FP, FN):

    # Precision
    precision = TP / (TP + FP) if (TP + FP) != 0 else 0

    # Recall
    recall = TP / (TP + FN) if (TP + FN) != 0 else 0

    # F1-Score
    f1_score = (
        2 * precision * recall / (precision + recall)
        if (precision + recall) != 0
        else 0
    )

    # Accuracy
    accuracy = (
        (TP + TN) / (TP + TN + FP + FN)
        if (TP + TN + FP + FN) != 0
        else 0
    )

    return accuracy, precision, recall, f1_score


# ---------------------------------------------------------
# Validation Metrics
# ---------------------------------------------------------

val_accuracy, val_precision, val_recall, val_f1 = calculate_metrics(
    TP_val, TN_val, FP_val, FN_val
)

print("=" * 55)
print("VALIDATION SET METRICS")
print("=" * 55)

print(f"Accuracy  : {val_accuracy:.4f}")
print(f"Precision : {val_precision:.4f}")
print(f"Recall    : {val_recall:.4f}")
print(f"F1-Score  : {val_f1:.4f}")


# ---------------------------------------------------------
# Test Metrics
# ---------------------------------------------------------

test_accuracy, test_precision, test_recall, test_f1 = calculate_metrics(
    TP_test, TN_test, FP_test, FN_test
)

print("\n" + "=" * 55)
print("TEST SET METRICS")
print("=" * 55)

print(f"Accuracy  : {test_accuracy:.4f}")
print(f"Precision : {test_precision:.4f}")
print(f"Recall    : {test_recall:.4f}")
print(f"F1-Score  : {test_f1:.4f}")
```

```python
# Task 10: ROC Curve, AUC and Final Performance Summary

import numpy as np
import matplotlib.pyplot as plt


# ---------------------------------------------------------
# 1. Calculate FPR and TPR for different thresholds
# ---------------------------------------------------------

thresholds = np.linspace(1, 0, 101)

fpr_values = []
tpr_values = []

for threshold in thresholds:

    # Convert probabilities into class labels
    y_pred_threshold = (y_test_prob >= threshold).astype(int)

    # Calculate TP, TN, FP, FN
    TP = np.sum((y_test == 1) & (y_pred_threshold == 1))
    TN = np.sum((y_test == 0) & (y_pred_threshold == 0))
    FP = np.sum((y_test == 0) & (y_pred_threshold == 1))
    FN = np.sum((y_test == 1) & (y_pred_threshold == 0))

    # False Positive Rate
    fpr = FP / (FP + TN) if (FP + TN) != 0 else 0

    # True Positive Rate
    tpr = TP / (TP + FN) if (TP + FN) != 0 else 0

    fpr_values.append(fpr)
    tpr_values.append(tpr)


fpr_values = np.array(fpr_values)
tpr_values = np.array(tpr_values)


# ---------------------------------------------------------
# 2. Calculate ROC-AUC using trapezoidal rule
# ---------------------------------------------------------

# Sort according to FPR
sort_index = np.argsort(fpr_values)

fpr_sorted = fpr_values[sort_index]
tpr_sorted = tpr_values[sort_index]

auc = np.trapezoid(tpr_sorted, fpr_sorted)


# ---------------------------------------------------------
# 3. Plot ROC Curve
# ---------------------------------------------------------

plt.figure(figsize=(8, 6))

plt.plot(
    fpr_sorted,
    tpr_sorted,
    label=f"Logistic Regression (AUC = {auc:.4f})"
)

# Random classifier line
plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--",
    label="Random Classifier"
)

plt.xlabel("False Positive Rate (FPR)")
plt.ylabel("True Positive Rate (TPR)")
plt.title("ROC Curve - Logistic Regression")
plt.legend()
plt.grid(True)

plt.show()


# ---------------------------------------------------------
# 4. Final Performance Summary
# ---------------------------------------------------------

print("=" * 60)
print("FINAL PERFORMANCE SUMMARY")
print("=" * 60)

print(f"Accuracy  : {test_accuracy:.4f}")
print(f"Precision : {test_precision:.4f}")
print(f"Recall    : {test_recall:.4f}")
print(f"F1-Score  : {test_f1:.4f}")
print(f"ROC-AUC   : {auc:.4f}")

print("\nConfusion Matrix:")
print(np.array([
    [TN_test, FN_test],
    [FP_test, TP_test]
]))

print("\nConfusion Matrix Interpretation:")
print(f"True Negative  (TN): {TN_test}")
print(f"False Negative (FN): {FN_test}")
print(f"False Positive (FP): {FP_test}")
print(f"True Positive  (TP): {TP_test}")
```
